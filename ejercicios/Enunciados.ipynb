## Problema de calor estacionario: 
Estudiamos el problema de transferencia de calor en estado estacionario en una barra (1D) $(0,1)$, que pierde calor hacia el ambiente.
$$-\Delta \varphi + \alpha u = f(x)\text{ en }\Omega,$$
junto con las condiciones frontera
$$\left. \varphi\right|_{\Gamma_D}=\beta \quad \left. \frac{\partial\varphi}{\partial \mathbf{n}}\right|_{\Gamma_N}=0,$$
con $\Gamma_D=\{0\}$, $\Gamma_N=\{1\}$.


Se considera: $\alpha = 1$, $\beta = 100$, $f(x)=1-x^2$.

## Membrana: 
Una membrana elástica $\Omega$ está sujeta a un soporte rígido plano $\Gamma$, y sometida a una fuerza $f(x)$. El desplazamiento vertical de la membrana $\varphi(x)$ se obtiene resolviendo la ecuación de Laplace
$$-\Delta \varphi = f(x)\text{ en }\Omega,$$
junto con las condiciones frontera
$$\left. \varphi\right|_{\Gamma_1}=0\quad \left. \frac{\partial\varphi}{\partial \mathbf{n}}\right|_{\Gamma_2}=g,$$
con $\partial \Omega = \Gamma_1\cup \Gamma_2$.

Se considera: $f(x,y)= -4(\cos(x^2+y^2-1) - (x^2+y^2)\sin(x^2+y^2-1))$, $g(x,y)=1.5$.

Etiquetas $\Gamma_1$: 2, $\Gamma_2$: 3
