```python
import nlopt
import numpy as np
```

```python

def objective_function(x, grad):
    if grad.size > 0:
        grad[0] = 2 * (x[0] - 5)  
        grad[1] = 2 * (x[1] + 10) 

    return (x[0] - 5)**2 + (x[1] + 10)**2
```

```python

n = 2
opt = nlopt.opt(nlopt.LD_MMA, n)

opt.set_min_objective(objective_function)

# Criterio de parada
opt.set_xtol_rel(1e-8)

# Punto de inicio
x0 = [0.0, 0.0]

try:
    x_opt = opt.optimize(x0)
    min_val = opt.last_optimum_value()
    
    print("--- Optimización con MMA (LD_MMA) ---")
    print(f"Mínimo encontrado en: x={x_opt[0]:.4f}, y={x_opt[1]:.4f}")
    print(f"Valor de f(x,y): {min_val:.6f}")
    print(f"Número de iteraciones: {opt.get_numevals()}")

except Exception as e:
    print(f"Error: {e}")
```
